```python
from math import gcd

class RSACripto:
    
    
    def __init__(self, p, q):
        
        self.p, self.q = p, q
        self.n = p*q
        self.phi = (p-1) * (q-1)
        
        self.e = 2
        
        while gcd(self.e, self.phi) != 1:
            self.e += 1
        
        self.d = 1
        
        while self.d * self.e % self.phi != 1:
            self.d += 1
        
    def crypt_pub(self, m):
        return pow(m, self.e, self.n)
    
    def decrypt_priv(self, c):
        return pow(c, self.d, self.n)


rsa = RSACripto(3121,3137)

rsa.e, rsa.d, rsa.n

for m in range(rsa.n):
    cript = rsa.crypt_pub(m)
    dec = rsa.decrypt_priv(cript)
    if m != dec:
        print(m, cript, dec)
    
print(gcd(rsa.p, rsa.n))
```
